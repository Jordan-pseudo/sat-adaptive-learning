```python
#To initialize a more efficient collection of items in python, we can use the built-in array structure.
import array
L = list(range(10))
La = array.array('i', L)
La
```

```python
#Using numpy array, we can put an python list as an input and construct a more efficient collection of data
import numpy as np
np.array(range(10))
```

```python
#We can also create a multidimensional list of items using the numpy array method
np.array([range(i, i + 3) for i in [2, 4, 6]])
```

```python
#Numpy also has some important methods
#to create an array of zeros\
np.zeros(3)
```

```python
#to create an array of ones
np.ones((3,3))
```

```python
#to create an array of filled with a constant
np.full((3,3), 3.14)
```

```python
#To create a linear sequence of numbers spaced in an equal interval
np.arange(0,20,2)
```

```python
#to create an array of values between two anchors equally spaced
np.linspace(0, 1, 5)
```

```python
#to create an array of random values between 0 and 1
np.random.random((3,3))
```

```python
#to create random values distributed noramlly with 0 mean and 1 standard deviation
np.random.normal(0,1,3)
```

```python
#to create a random list of integers between two values (0 & 10, for instance), we use the folliwng method
np.random.randint(0, 10, (4, 5))
```

```python
#to create a identity matrix
np.eye(4)
```

```python
#to create an unitialized array
np.empty(3)
```

```python
#To analyze and study different attributes of a numpy array, we can use some of the following methods:
rng = np.random.default_rng(seed = 43)

x1 = rng.integers(10, size = 6)
x2 = rng.integers(10, size = (3, 5))
x3 = rng.integers(10, size = (3, 5, 6))

print("size of x2: " , x2.ndim)
print("size of x2: " , x2.shape)
print("size of x2: " , x2.size)
print("size of x2: " , x2.dtype)
```

```python
#accessing an element in a numpy array is similar to how we access elements in a python list
print(x2[0,1])
```

```python
#to concatenate np arrays, we can use the following notation
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
np.concatenate([x, y])
```

```python
grid = np.array([[1, 2, 3],
                 [4, 5, 6]])
np.concatenate([grid, grid])
```

```python
np.concatenate([grid, grid], axis = 1)
```
